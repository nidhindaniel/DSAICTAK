```python

sales_data = []
for i in range(11):
    order = {
        "order_id": int(input("Enter the ORDER_ID: ")),
        "customer": str(input("Enter the Customer Name: ")),
        "item": str(input("Enter the Item: ")),
        "quantity": int(input("Enter the Quantity: ")),
        "price_per_item": float(input("Enter the Price per Item: ")),
        "order_date": str(input("Enter the Order Date (YYYY-MM-DD): "))
    }
    sales_data.append(order)  

print("All Sales Records:", sales_data)
```

```python

def total_revenue(orders):
   
    total = sum(order["quantity"] * order["price_per_item"] for order in orders)
    return total


grand_total = total_revenue(sales_data)
print(f"Total Revenue: {grand_total}")
```

```python
for order in sales_data:
    order_total = order["quantity"] * order["price_per_item"]
    if order_total > 2000:
        print(f"Order {order['order_id']} by {order['customer']}: ₹{order_total:,.2f}")
        
```

```python
item_quantities = {}
for order in sales_data:
    item = order["item"]
    item_quantities[item] = item_quantities.get(item, 0) + order["quantity"]
    best_selling_item = max(item_quantities, key=item_quantities.get)
print(f"\nHighest Quantity Sold: {best_selling_item} ({item_quantities[best_selling_item]} units)")
```

```python
def apply_discount(orders, discount_rate):
    discounted_orders = []
    for order in orders:
        new_order = order.copy()
        new_order["price_per_item"] = new_order["price_per_item"] * (1 - discount_rate)
        discounted_orders.append(new_order)
    return discounted_orders


discounted_sales_data = apply_discount(sales_data, 0.10)

print("Original Price:", sales_data[0]["price_per_item"])
print("Discounted Price:", discounted_sales_data[0]["price_per_item"])
```

```python

store_country = "India" #global

def calculate_tax(order):
    tax_rate = 0.18 #local
    print(f"Calculating tax for store location: {store_country}")
    return order["price_per_item"] * tax_rate


sample_order = {"price_per_item": 1000.0}
print("Tax amount:", calculate_tax(sample_order))

print("Global location:", store_country)

```

```python
class OrderIterator:

    def __init__(self, orders):
        self.orders = orders

    def __iter__(self):
        for order in self.orders:
            yield order

for order in OrderIterator(sales_data):
    print(f"Order #{order['order_id']} - {order['customer']}")
```

```python
orders_by_customer = {
    customer: [order for order in sales_data if order["customer"] == customer]
    for customer in {order["customer"] for order in sales_data}
}

for customer, orders in orders_by_customer.items():
    print(f"\nCustomer: {customer}")
    for order in orders:
        print(f"  - Order #{order['order_id']}: {order['item']} (Qty: {order['quantity']})")
```
