## 1. Dataset Overview

```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.preprocessing import StandardScaler

path = Path("House_Pricing.csv")
if not path.exists():
    path = Path("/mnt/user-data/uploads/House_Pricing.csv")

df = pd.read_csv(path)
df.head()
```

## 2. Duplicate Removal

```python
print(df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)

dup_cols = df.columns[df.T.duplicated()].tolist()
print(dup_cols)
df = df.drop(columns=dup_cols)
df.shape
```

```python
df["Date House was Sold"] = pd.to_datetime(df["Date House was Sold"], format="%d %B %Y")
df["Sale Year"] = df["Date House was Sold"].dt.year
df["Sale Month"] = df["Date House was Sold"].dt.month
df = df.drop(columns=["ID", "Date House was Sold"])
```

## 3. Handling Missing Values
- `Sale Price` (target): rows with missing values dropped
- `No of Times Visited`: missing means the house was never visited, filled with "Never" (the mode would add visits that never happened)
- `Zipcode`: mode
- All other numerical columns: median (distributions are right skewed)

```python
df = df.dropna(subset=["Sale Price"]).reset_index(drop=True)

df["No of Times Visited"] = df["No of Times Visited"].fillna("Never")
df["Zipcode"] = df["Zipcode"].fillna(df["Zipcode"].mode()[0]).astype(int).astype(str)

for col in df.select_dtypes("number").columns:
    df[col] = df[col].fillna(df[col].median())

df.isnull().sum().sum()
```

## 4. Scaling Numerical Variables

```python
num_cols = [c for c in df.select_dtypes("number").columns if c != "Sale Price"]
raw_num = df[num_cols].copy()

scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
df[num_cols].describe().round(2)
```

## 5. Encoding Categorical Variables

```python
cat_cols = df.select_dtypes(exclude="number").columns.tolist()
print(cat_cols)

ordinal = {
    "Condition of the House": ["Bad", "Okay", "Fair", "Good", "Excellent"],
    "No of Times Visited": ["Never", "Once", "Twice", "Thrice", "Four"],
}
for col, order in ordinal.items():
    df[col] = df[col].map({v: i for i, v in enumerate(order)})

nominal = [c for c in cat_cols if c not in ordinal]
df = pd.get_dummies(df, columns=nominal, drop_first=True, dtype=int)
df.shape
```

## 6. Outlier Removal

```python
iqr_cols = ["Sale Price", "No of Bedrooms", "No of Bathrooms", "Flat Area (in Sqft)",
            "Lot Area (in Sqft)", "Area of the House from Basement (in Sqft)",
            "Living Area after Renovation (in Sqft)", "Lot Area after Renovation (in Sqft)"]

raw_all = raw_num.copy()
raw_all["Sale Price"] = df["Sale Price"]

mask = pd.Series(False, index=df.index)
for col in iqr_cols:
    q1, q3 = raw_all[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    out = (raw_all[col] < q1 - 1.5 * iqr) | (raw_all[col] > q3 + 1.5 * iqr)
    print(col, out.sum())
    mask |= out

print(mask.sum())
```

```python
df = df.loc[~mask].copy()
df[num_cols] = StandardScaler().fit_transform(raw_num.loc[df.index])
df = df.reset_index(drop=True)
print(df.shape)
df.head()
```

```python
df.to_csv("House_Pricing_cleaned.csv", index=False)
```
